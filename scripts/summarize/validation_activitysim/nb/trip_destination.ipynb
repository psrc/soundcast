## district flow

- \(model trips - survey trips\) / survey trips

```python
# get district names for home and work locations
dist_name = {1: '1: Suburban Snohomish',
             2: '2: Everett-Lynwood-Edmonds',
             3: '3: North Seattle-Shoreline',
             4: '4: Seattle CBD',
             5: '5: West-South Seattle',
             6: '6: East Side',
             7: '7: Renton-FedWay-Kent',
             8: '8: Tacoma',
             9: '9: Kitsap',
             10: '10: South Pierce'}
# get maz to district lookup
geog_lookup_maz_district = util.read_sqlite_db(input_config, summary_config, "SELECT * FROM parcel_2023_geography").\
    drop_duplicates(subset=['maz_id'], keep='first').\
    sort_values(by=['maz_id'])
maz_district_lookup = dict(zip(geog_lookup_maz_district['maz_id'], geog_lookup_maz_district['District']))

trip_data_dist = trip_data.copy()
trip_data_dist['origin_district'] = trip_data_dist['origin'].map(maz_district_lookup)
trip_data_dist['dest_district'] = trip_data_dist['destination'].map(maz_district_lookup)

trip_data_dist['origin_dist_name'] = trip_data_dist['origin_district'].map(dist_name)
trip_data_dist['dest_dist_name'] = trip_data_dist['dest_district'].map(dist_name)
```

```python
from pandas.api.types import CategoricalDtype

# model district flow
df_model = trip_data_dist.\
    groupby(['source','origin_dist_name','dest_dist_name'], observed=True)['trip_weight'].sum().\
    reset_index()

df_cols = ['origin_dist_name','dest_dist_name','person_weight','source']

dist_type = CategoricalDtype(categories=dist_name.values(), ordered=True)

df_model['origin_dist_name'] = df_model['origin_dist_name'].astype(dist_type)
df_model['dest_dist_name'] = df_model['dest_dist_name'].astype(dist_type)
```

```python
# get percent difference between model and LEHD
source_list = df_model['source'].unique().tolist()
source_list.remove('model')
```

```python
for sc in source_list:
    df_plot = pd.pivot_table(df_model, values='trip_weight', index=['origin_dist_name','dest_dist_name'],
                            columns=['source']).reset_index()
    df_plot['pct_diff'] = (df_plot['model'] - df_plot[sc]) / df_plot[sc]

    df_plot2 = pd.pivot_table(df_plot, values='pct_diff', index=['origin_dist_name'],
                            columns=['dest_dist_name']).reset_index()

    # Count survey samples per origin/dest district pair
    df_count = trip_data_dist[trip_data_dist['source'] == sc].\
        groupby(['origin_dist_name', 'dest_dist_name'], observed=True).size().\
        reset_index(name='n_survey')
    df_count['origin_dist_name'] = df_count['origin_dist_name'].astype(dist_type)
    df_count['dest_dist_name'] = df_count['dest_dist_name'].astype(dist_type)
    df_n = pd.pivot_table(df_count, values='n_survey', index=['origin_dist_name'],
                        columns=['dest_dist_name'], aggfunc='sum')

    # Build display strings: "xx% (n)"
    pct_indexed = df_plot2.set_index('origin_dist_name')
    fmt_df = pct_indexed.copy().astype(object)
    for col in pct_indexed.columns:
        for idx in pct_indexed.index:
            pct_val = pct_indexed.loc[idx, col]
            if not np.isfinite(pct_val):
                fmt_df.loc[idx, col] = 'N/A'
            else:
                try:
                    n_val = int(df_n.loc[idx, col])
                    fmt_df.loc[idx, col] = f"{pct_val:.1%} ({n_val:,})"
                except (KeyError, ValueError):
                    fmt_df.loc[idx, col] = f"{pct_val:.1%}"

    display(fmt_df.style.set_caption(f"District distribution: {sc}"))
```
