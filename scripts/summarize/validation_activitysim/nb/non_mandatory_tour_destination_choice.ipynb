## non-mandatory tour distance

```python
util.plot_share_barchart(non_m_tour_data, 'tour_weight', 'tour_distance_bin', title="share of non-mandatory tours by distance bin")
```

```python
df_plot = non_m_tour_data.groupby(['source','tour_distance_bin_60mi'], observed=True)['tour_weight'].sum().reset_index()
df_plot['percentage'] = df_plot.groupby(['source'], group_keys=False, observed=True)['tour_weight']. \
    apply(lambda x: x / float(x.sum()))

fig2 = px.line(df_plot, x='tour_distance_bin_60mi', y="percentage", color="source", title="tour distance")
fig2.for_each_annotation(lambda a: a.update(text = a.text.split("=")[-1]))
fig2.update_layout(height=400, width=700, font=dict(size=11), yaxis=dict(tickformat=".1%"))
fig2.show()
```
