## Network Counts

- model results: `outputs/network/network_results.csv`
- observed data from `observed_jblm_counts`

```python

# summary data location
network_results = pd.read_csv(Path(summary_config['sc_run_path']) / 'outputs/network/network_results.csv')
observed_jblm_counts = util.read_sqlite_db(input_config, summary_config, "SELECT * FROM observed_jblm_counts")
```

```python
def load_counts(df_obs, year):

    df_obs = df_obs[df_obs["Year"]==year]
    df_obs = df_obs[~df_obs['Observed'].isnull()]

    df = network_results[["@countid","@tveh"]].groupby("@countid").sum().reset_index()
    df = df.merge(df_obs, left_on="@countid", right_on="CountID")

    df = df.groupby("CountID").sum()[['@tveh','Observed','Location']]
    df.loc["Total",:] = df[['@tveh','Observed']].sum(axis=0)
    df.fillna("All JBLM", inplace=True)

    df.rename(columns={'@tveh': 'Model', 'Observed': f'Observed {year}'}, inplace=True)

    pd.options.display.float_format = '{:0,.0f}'.format
    df[['Location','Model',f'Observed {year}']]

    return df
```

```python
min_year = observed_jblm_counts['Year'].min()
max_year = observed_jblm_counts['Year'].max()

df_min = load_counts(observed_jblm_counts, min_year)
df_max = load_counts(observed_jblm_counts, max_year)

df = df_min.merge(df_max[[f'Observed {max_year}']], left_index=True, right_index=True)
df[['Location','Model',f'Observed {min_year}',f'Observed {max_year}']]
```

## Trips from create_ixxi_work_trips.py
### Vehicle trips to/from JBLM zones by origin/destination 

```python
jblm_taz_list = util.read_sqlite_db(input_config, summary_config, "SELECT * FROM psrc_zones where jblm=1")['taz'].to_list()
```

```python
# Military work trips are added as ODs in ixxi_work_trips.h5

def get_external_trips(myh5, NUM_ZONES):

    # Read trip table and get indeces
    # Read table data for each mode and combine into single matrix
    
    # Note: we are assuming ordinal indexing for externals, do not use these to select zone IDs outside of internal zones

    trips = np.zeros((NUM_ZONES, NUM_ZONES), dtype=np.float64)
    for mode in ['sov','hov2','hov3']:
        trips += myh5[mode][:]
    df = pd.DataFrame(
        trips, 
        columns=range(1,NUM_ZONES+1), 
        index=range(1,NUM_ZONES+1)
        )
    
    return df
```

```python
# get number of zones
tod = list(summary_config['tod_lookup'].keys())[0]
myh5 = h5py.File(root_dir / 'outputs/supplemental' / f'external_work_{tod}.h5', 'r')
NUM_ZONES = len(myh5['sov'][:])

# external work trips
external_work_trips = pd.DataFrame(
    np.zeros([NUM_ZONES, NUM_ZONES]),
    columns=range(1,NUM_ZONES+1), 
    index=range(1,NUM_ZONES+1)
    )

for work_h5 in summary_config['tod_lookup'].keys():
    
    myh5 = h5py.File(root_dir / 'outputs/supplemental' / f'external_work_{work_h5}.h5', 'r')
    df = get_external_trips(myh5, NUM_ZONES)
    external_work_trips = external_work_trips + df

# external non-work trips
myh5 = h5py.File(f"{root_dir}/outputs/supplemental/external_non_work.h5", 'r')
external_non_work_trips = get_external_trips(myh5, NUM_ZONES)
```

```python

# Summarize vehicle trips between all zones and JBLM zones
work_trips_to_jblm = external_work_trips[jblm_taz_list].sum().rename("work trips to JBLM")
work_trips_from_jblm = external_work_trips.loc[jblm_taz_list].sum(axis=1).rename("work trips from JBLM")
non_work_trips_to_jblm = external_non_work_trips[jblm_taz_list].sum().rename("non-work trips to JBLM")
non_work_trips_from_jblm = external_non_work_trips.loc[jblm_taz_list].sum(axis=1).rename("non-work trips from JBLM")

# Combine and compute totals
df = pd.concat([work_trips_to_jblm, work_trips_from_jblm, non_work_trips_to_jblm, non_work_trips_from_jblm], axis=1).fillna(0)
df["total work trips"] = df["work trips to JBLM"] + df["work trips from JBLM"]
df["total non-work trips"] = df["non-work trips to JBLM"] + df["non-work trips from JBLM"]

df
```

```python
total_non_work_trips = int(external_non_work_trips.sum().sum())
print(f"Total non-work external trips in region: {total_non_work_trips:,}")

# Non work trips to/from externals
external_non_work_trips_ext = int(external_non_work_trips.loc[3700:].sum().sum())
# external_non_work_trips.loc[3700:].sum().sum()
print(f"External -> Internal non-work external trips in region: {external_non_work_trips_ext:,}")

# Non work trips to/from JBLM
external_non_work_trips_ext = int(external_non_work_trips[3700:].sum().sum())
# external_non_work_trips.loc[3700:].sum().sum()
print(f"Internal -> External non-work external trips in region: {external_non_work_trips_ext:,}")

# Internal -> Internal non-work trips
external_non_work_trips_int = int(external_non_work_trips.loc[1:3700,1:3700].sum().sum())
print(f"Internal -> Internal non-work external trips in region: {external_non_work_trips_ext:,}")
```

## Daysim IXXI File

This file in generated in create_ixxi_work_trips.py and specifies what share of each zone to exclude from DaySim Work Location model

Within JBLM area, Jobs_XIFrac should be 1 because we are using the exogenous data for work flows in these zones
If Jobs_XIFrac == 1 then Daysim will not send workers to jobs in these zones

Worker_IXFrac is not equal to 1 because workers in the synthetic population for these areas 
may have jobs outside the base area (family members of enlisted)

Compare Military Jobs by Zone before create_ixxi_work_trips.py and afterwards

JBLM Zones should have 0 total employment in output file

```python
df_before = df_parcel_input[df_parcel_input['taz_p'].isin(parcels_military.Zone)].groupby('taz_p')['emptot_p'].sum().reset_index().sort_values(by='emptot_p', ascending=False)
df_after = df_parcel_output[df_parcel_output['taz_p'].isin(parcels_military.Zone)].groupby('taz_p')['emptot_p'].sum().reset_index().sort_values(by='emptot_p', ascending=False)

df = df_before.merge(df_after, left_on='taz_p', right_on='taz_p', how='outer')
df = df.rename(columns={'taz_p': 'taz', 'emptot_p_x': 'emptot_p_input', 'emptot_p_y': 'emptot_p_output'})

df = df.merge(parcels_military, left_on='taz', right_on='Zone', how='outer')
df.drop(columns=['Zone'], inplace=True)

# set index to taz
df = df.set_index('taz')

df_out = df.copy()
df_out.loc['Total'] = df.sum(numeric_only=True)
# # Show results with commas and as integer
# for col in df.columns:
#     df[col] = df[col].apply(lambda x: f"{int(x):,}")
pd.options.display.float_format = '{:0,.0f}'.format

# Show JBLM Zones
df_out[df_out.index.isin(jblm_taz_list)][['emptot_p_input','emptot_p_output']].astype('float')
```
